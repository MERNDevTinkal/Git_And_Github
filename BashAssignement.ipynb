


```markdown
 Task 1: File Management Script

Write a Bash script that:
- Creates a directory named "backup" in the user's home directory
- Copies all .txt files from the current directory into the "backup" directory
- Appends the current date and time to the filenames of the copied files.

Answer >>>>>>>>

```bash
#!/bin/bash

mkdir -p ~/backup

timestamp=$(date +"%Y-%m-%d_%H-%M-%S")

for file in *.txt; do
    [ -e "$file" ] || continue
    base=$(basename "$file" .txt)
    cp "$file" ~/backup/"${base}_$timestamp.txt"
done

echo " Backup complete. Files saved in ~/backup"
```

```markdown
 Task 2: System Health Check

Create a script that:
- Checks the system’s CPU and memory usage
- Reports if the CPU usage is above 80% or if the available memory is below 20%
- Logs the results to a file named system_health.log.

```bash
#!/bin/bash

log_file="system_health.log"
timestamp=$(date "+%Y-%m-%d %H:%M:%S")

cpu_usage=$(top -bn1 | grep "Cpu(s)" | awk '{print 100 - $8}')
cpu_usage_int=${cpu_usage%.*}

read total used free <<< $(free -m | awk '/Mem:/ {print $2, $3, $7}')
mem_available_percent=$(( free * 100 / total ))

echo "[$timestamp] CPU: $cpu_usage% | Free Memory: $mem_available_percent%" >> $log_file

if [ "$cpu_usage_int" -gt 80 ]; then
    echo "Warning: CPU usage is above 80%" >> $log_file
fi

if [ "$mem_available_percent" -lt 20 ]; then
    echo "Warning: Available memory is below 20%" >> $log_file
fi

echo " System check logged to $log_file"
```

```markdown
 Task 3: User Account Management

Write a script that:
- Reads a list of usernames from a file (e.g., user_list.txt)
- Creates a new user for each username
- Generates a random password for each user and saves the username and password to a file named credentials.txt

```bash
#!/bin/bash

input="user_list.txt"
output="credentials.txt"
> $output

while read username; do
    useradd -m "$username"
    password=$(openssl rand -base64 12)
    echo "$username:$password" | chpasswd
    echo "$username : $password" >> $output
done < "$input"

echo " Users created and saved to $output"
```

```markdown
 Task 4: Automated Backup

Create a script that:
- Takes a directory path as input from the user
- Compresses the directory into a .tar.gz file
- Saves the compressed file with a name that includes the current date (e.g., backup_2023-08-20.tar.gz).


```bash
#!/bin/bash

read -p "Enter the directory path: " dir

if [ ! -d "$dir" ]; then
    echo " Directory not found!"
    exit 1
fi

date_stamp=$(date +"%Y-%m-%d")
output="backup_$date_stamp.tar.gz"

tar -czf "$output" "$dir"

echo " Backup created: $output"
```

```markdown
  Task 5: Simple To-Do List

Create a Bash script that:
- Implements a simple command-line to-do list
- Allows the user to add tasks, view tasks, and remove tasks
- Saves the tasks to a file (e.g., todo.txt).


```bash
#!/bin/bash

file="todo.txt"

echo "1. Add Task"
echo "2. View Tasks"
echo "3. Remove Task"
read -p "Choose an option: " choice

case $choice in
    1)
        read -p "Enter task: " task
        echo "$task" >> $file
        echo " Task added."
        ;;
    2)
        echo " Your To-Do List:"
        nl $file
        ;;
    3)
        nl $file
        read -p "Enter task number to remove: " num
        sed -i "${num}d" $file
        echo " Task removed."
        ;;
    *)
        echo " Invalid option."
        ;;
esac
```

```markdown
 Task 6: Automated Software Installation

Write a script that:
- Reads a list of software package names from a file (e.g., packages.txt)
- Installs each package using the appropriate package manager (apt, yum, etc.)
- Logs the installation status of each package.


```bash
#!/bin/bash

log="install_log.txt"
input="packages.txt"

while read package; do
    echo "Installing $package..."
    if sudo apt install -y "$package" >> $log 2>&1; then
        echo "$package installed successfully" >> $log
    else
        echo "Failed to install $package" >> $log
    fi
done < "$input"

echo " Installation process completed. Check $log for details."
```

```markdown
  Task 7: Text File Processing

Create a script that:
- Takes a text file as input
- Counts and displays the number of lines, words, and characters in the file
- Finds and displays the longest word in the file.


```bash
#!/bin/bash

read -p "Enter the text file name: " file

if [ ! -f "$file" ]; then
    echo " File not found."
    exit 1
fi

lines=$(wc -l < "$file")
words=$(wc -w < "$file")
chars=$(wc -m < "$file")
longest=$(tr -s '[:space:]' '\n' < "$file" | awk '{ if (length > max) { max = length; word = $0 } } END { print word }')

echo " File: $file"
echo " Lines: $lines"
echo " Words: $words"
echo " Characters: $chars"
echo " Longest word: $longest"
```
