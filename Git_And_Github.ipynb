### 2 Tasks to Be Performed:

1. Create a Git working directory with feature1.txt and feature2.txt in the master branch
2. Create 3 branches: develop, feature1, and feature2
3. In develop branch, create develop.txt (but do not stage or commit)
4. Stash this file and checkout to feature1 branch
5. In feature1 branch, create new.txt, stage and commit it
6. Checkout back to develop, unstash the file, and commit
7. Submit all Git commands used

Answer --->>>

```bash
# Step 1: Create working directory and initialize Git
mkdir GitBranchTask
cd GitBranchTask
git init

# Create files in master branch
touch feature1.txt feature2.txt
git add feature1.txt feature2.txt
git commit -m "commit with feature1 and feature2 files"

# Step 2: Create branches
git branch develop
git branch feature1
git branch feature2

# Step 3: Checkout to develop and create develop.txt
git checkout develop
touch develop.txt

# Step 4: Stash the file (untracked stash)
git stash --include-untracked

# Step 5: Switch to feature1 and commit new.txt
git checkout feature1
touch new.txt
git add new.txt
git commit -m "Added new.txt in feature1 branch"

# Step 6: Back to develop and unstash
git checkout develop
git stash pop

# Commit develop.txt
git add develop.txt
git commit -m "Added develop.txt in develop branch"
```

### 4. Tasks to Be Performed:

1. Put master.txt on master branch, stage and commit
2. Create 3 branches: public 1, public 2 and private
3. Put public1.txt on public 1 branch, stage and commit
4. Merge public 1 on master branch
5. Merge public 2 on master branch
6. Edit master.txt on private branch, stage and commit
7. Update branch public 1 and public 2 with new master code in private
8. Update new master code on master
9. Finally update all the code on the private branch

Answer --->>>

```bash
# 1. Put master.txt on master branch
echo "This is master file" > master.txt
git add master.txt
git commit -m "Add master.txt on master branch"

# 2. Create branches
git branch "public 1"
git branch "public 2"
git branch private

# 3. Put public1.txt on public 1 branch
git checkout "public 1"
echo "This is public1 file" > public1.txt
git add public1.txt
git commit -m "Add public1.txt on public 1 branch"

# 4. Merge public 1 on master
git checkout master
git merge "public 1"

# 5. Merge public 2 on master
git merge "public 2"

# 6. Edit master.txt on private branch
git checkout private
echo "Updated by private branch" >> master.txt
git add master.txt
git commit -m "Edited master.txt in private branch"

# 7. Update public branches with private changes
git checkout "public 1"
git merge private

git checkout "public 2"
git merge private

# 8. Update master with private changes
git checkout master
git merge private

# 9. Final update on private
git checkout private
git merge master
```

### 5. Tasks to Be Performed:

1. Create a Git Flow workflow architecture on Git  
2. Create all the required branches  
3. Starting from feature branch, push to master following the architecture  
4. Push urgent.txt on master using hotfix  

**Answer --->>>**  

```bash
# 1. Initialize Git Flow (creates master, develop, and supporting branch structure)
git flow init -d

# 2. Create and work on a feature branch
git flow feature start new-feature
touch feature-file.txt
git add feature-file.txt
git commit -m "Added new feature implementation"

# 3. Complete feature workflow (feature → develop → release → master)
git flow feature finish new-feature
git flow release start v1.0
git flow release finish v1.0

# 4. Emergency hotfix procedure
git flow hotfix start urgent-fix
echo "Critical production fix" > urgent.txt
git add urgent.txt
git commit -m "Added urgent hotfix file"
git flow hotfix finish urgent-fix

# Push all branches to remote
git push origin --all
git push origin --tags
```
